```python
import pandas as pd
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({"marks":[80,70,None,90,60],"Attendence":[90,None,85,80,75],"Department":["CSE","ECE","EEE",None,"CSE"]})
```

```python
#mean imputation
df["Attendence"]=df["Attendence"].fillna(df["Attendence"].median())
```

```python
#median imputation
df["marks"]=df["marks"].fillna(df["marks"].mean())
```

```python
#forwrd fill
df["Attendence"]=df["Attendence"].ffill()

#fill missing categorical value using mode
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
#label coding
encoder=LabelEncoder()
df["Department"]=encoder.fit_transform(df["Department"])
print(df)
```
